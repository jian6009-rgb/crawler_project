```python
import requests
```

```python
url = "https://www.kugou.com/yy/rank/home/1-8888.html?from=rank" #原本想选主界面，但因为主界面元素太多太杂，所以选了榜单的url

headers = {"User-Agent":'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36 SLBrowser/9.0.8.7271 SLBChan/115 SLBVPV/64-bit'}

response = requests.get(url,headers=headers,timeout=10)
```

```python
#检查是否正确
print(response.status_code)
print(response.url)
print(response.headers.get("Content-Type"))
print(len(response.content))
```

```python
html = response.text
#检查text是不是正确
print(html[:500])
```

```python
example1 = "山风山风等等我" #榜单第一首歌，大致看看信息
print(example1 in html)
print(html.find(example1))
```

```python
position = html.find(example1)

print(html[position - 200:position + 1500])#榜单第一首歌到第二首哥的信息，完整看看每一首歌html是怎么写的
```

```python
soup = BeautifulSoup(html, "html.parser")
song_tags = soup.select(".pc_temp_songlist li")
songs = []
for song_tag in song_tags:
    
    every_song_tag = song_tag.select_one("a.pc_temp_songname") #找到有pc_temp_songname的a
    
    full_title = song_tag.get("title") #select 找文字，get 找属性
    singer_name, song_name = full_title.split(" - ", 1) #title格式：万海东 - 山风山风等等我
    rank = song_tag.select_one(".pc_temp_num").get_text(strip=True)
    duration = song_tag.select_one(".pc_temp_time").get_text(strip=True)
    song_url = every_song_tag.get("href")

    song = {"rank": rank,"song_name": song_name,"singer_name": singer_name,"duration": duration,"song_url": song_url}
    songs.append(song)
    
songs[21]
```
